```python
df = pd.read_csv('/content/drive/MyDrive/fifa_eda_stats.csv')
```

```python
df.drop('ID', axis=1, inplace=True) # removing irrelevant information!
```

```python
df
```

```python
df.fillna(method='ffill', inplace=True) # filling in any values that are empty to help with processing!
```

```python
print(df.head())
```

```python
df['Wage'] = df['Wage'].str.replace('€', '')
df['Value'] = df['Value'].str.replace('€', '')
df['Wage'] = df['Wage'].str.replace('M', 'e6')
df['Wage'] = df['Wage'].str.replace('K', 'e3')
df['Value'] = df['Value'].str.replace('K', 'e3')
df['Value'] = df['Value'].str.replace('M', 'e6')
# making sure that all of the values are strictly numerical and can be read easily!
```

```python
df['Value'] = df['Value'].astype(float)
df['Wage'] = df['Wage'].astype(float)
# changing the numbers into floats to keep their integrity!
```

```python
values_wages = df[['Value', 'Wage']]
```

## Pre-Processing the data!

Let's Scale our Data!

```python
scaler = StandardScaler()
df_scaled = scaler.fit_transform(values_wages)
```

```python
df_norm = pd.DataFrame(df_scaled)
```

```python
plt.figure(figsize =(20, 8))
plt.title('Visualizing the data')
linked = linkage(df_scaled, method='ward')
dendrogram(linked, p=10, truncate_mode='lastp',
               orientation='top',
               distance_sort='descending',
               show_leaf_counts=True)
plt.xlabel('Player Value Clusters')
plt.ylabel('Player Wage')
plt.show()
```
